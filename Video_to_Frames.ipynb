```python
# i.e if video of duration 30 seconds, saves 10 frame per second = 300 frames saved in total
SAVING_FRAMES_PER_SECOND = 30
```

```python
def format_timedelta(td):
    """Utility function to format timedelta objects in a cool way (e.g 00:00:20.05)
    omitting microseconds and retaining milliseconds"""
    result = str(td)
    try:
        result, ms = result.split(".")
    except ValueError:
        return (result + ".00").replace(":", "-")
    ms = int(ms)
    ms = round(ms / 1e4)
    return f"{result}.{ms:02}".replace(":", "-")


def get_saving_frames_durations(cap, saving_fps):
    """A function that returns the list of durations where to save the frames"""
    s = []
    print("cv2.CAP_PROP_FRAME_COUNT: ", cv2.CAP_PROP_FRAME_COUNT)
    # get the clip duration by dividing number of frames by the number of frames per second
    clip_duration = cap.get(cv2.CAP_PROP_FRAME_COUNT) / cap.get(cv2.CAP_PROP_FPS)
    # use np.arange() to make floating-point steps
    for i in np.arange(0, clip_duration, 1 / saving_fps):
        s.append(i)
    return s
```

```python
def main(video_file):
    filename, _ = os.path.splitext(video_file)
    filename = "/content/drive/MyDrive/Nasa_Flame/Green_Hot_ver2" #+ "-opencv"
    # make a folder by the name of the video file
    if not os.path.isdir(filename):
        os.mkdir(filename)
    # read the video file
    cap = cv2.VideoCapture(video_file)
    # get the FPS of the video
    fps = cap.get(cv2.CAP_PROP_FPS)
    print("FPS of Video: ", fps)
    # if the SAVING_FRAMES_PER_SECOND is above video FPS, then set it to FPS (as maximum)
    saving_frames_per_second = min(fps, SAVING_FRAMES_PER_SECOND)
    # get the list of duration spots to save
    saving_frames_durations = get_saving_frames_durations(cap, saving_frames_per_second)
    print(len(saving_frames_durations))
    # start the loop
    count = 0
    for i in tqdm(range(int(cap.get(cv2.CAP_PROP_FRAME_COUNT)))): #while True:
        is_read, frame = cap.read()
        #plt.imshow(frame)
        if not is_read:
            print("NOT READ")
            # break out of the loop if there are no frames to read
            break
        # get the duration by dividing the frame count by the FPS
        frame_duration = count / fps
        try:
            # get the earliest duration to save
            closest_duration = saving_frames_durations[0]
        except IndexError:
            # the list is empty, all duration frames were saved
            break
        #if frame_duration >= closest_duration:
        # if closest duration is less than or equals the frame duration,
        # then save the frame
        frame_duration_formatted = format_timedelta(timedelta(seconds=closest_duration))
        cv2.imwrite(os.path.join(filename, f"frame{frame_duration_formatted}.jpg"), frame)
        # drop the duration spot from the list, since this duration spot is already saved
        try:
            saving_frames_durations.pop(0)
        except IndexError:
            pass
        # increment the frame count
        count += 1
```

```python
main("/content/drive/MyDrive/Nasa_Flame/4-GreenHot.mov")
```
